```python
import pandas as pd

df = pd.read_csv("../data/sensor_data.csv")

df.head()
```

```python


df.shape
```

```python
# Temperature
df["temp_rolling_mean_20"] = df["temperature"].rolling(window=20).mean()
df["temp_rolling_std_20"] = df["temperature"].rolling(window=20).std()

# Vibration
df["vib_rolling_mean_20"] = df["vibration"].rolling(window=20).mean()
df["vib_rolling_std_20"] = df["vibration"].rolling(window=20).std()

# Pressure
df["pressure_rolling_mean_20"] = df["pressure"].rolling(window=20).mean()
df["pressure_rolling_std_20"] = df["pressure"].rolling(window=20).std()
```

```python
df.shape
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
df.shape
```

```python
features = [
    "temperature",
    "vibration",
    "pressure",
    "temp_rolling_mean_20",
    "temp_rolling_std_20",
    "vib_rolling_mean_20",
    "vib_rolling_std_20",
    "pressure_rolling_mean_20",
    "pressure_rolling_std_20"
]

X = df[features]
y = df["failure_warning"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)

X.head()
```

```python
split_index = int(len(X) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\nTrain warnings:")
print(y_train.value_counts())

print("\nTest warnings:")
print(y_test.value_counts())
```
